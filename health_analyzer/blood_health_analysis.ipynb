```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
with open("blood_analysis.txt", "r") as f:
    blood_report = f.read()

    print(blood_report [:200])
```

```python
extraction_prompt = f"""
You are a medical data extraction assistant.

From the blood report below, extract all test values and classify each one as HIGH, LOW, or NORMAL based
on reference ranges provided in th report.

Format your response as:
- Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range
Blood Report:
{blood_report}
"""
extraction_response = llm.invoke(extraction_prompt)
extracted_values = extraction_response.text

print("=== STAGE 1: EXTRACTED VALUES ===")
print(extracted_values)
```

```python
diet_prompt = f"""
You are a clinical nutritionist specializing in African dietary habits.

Based on the blood work analysis below, write:
1. A short health summary in 3 lines explaining the patient's condition in simple language.
2. A shorth practical South African diet plan having only two sections (1) Foods to avoid, (2) Food to eat more often.
   Do not include any other sections in the diet plan.
Blood Work Analyses:
{extracted_values}
"""
diet_response = llm.invoke(diet_prompt)
diet_plan = diet_response.text

print("=== STAGE 2: HEALTH SUMMARY & DIET PLAN ===")
print(diet_plan)
```
