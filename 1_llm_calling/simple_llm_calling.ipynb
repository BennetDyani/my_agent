```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assisstant that answers in short one line, less than 50 words"],
    ["human", "How many moons does Jupiter have?"],
    ])
```

```python
prompt = "Write a single tagline for a quick cofee shop. Only output the tagline, nothing else."
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

response = llm.invoke(prompt)
print(response.text)
```
